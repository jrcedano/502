```python
number1 = float(input("Enter Number 1: "))
number2 = float(input("Enter Number 2:" ))
operation = input("Select your operation (+; -; *; /): ")

if operation == "+":
    answer = number1 + number2

elif operation == "-":
    answer = number1 - number2

elif operation == "*":
    answer = number1 * number2

elif operation == "/":
    if number2 == 0:
        answer = "Cannot divide by 0"
    else:
        answer = number1 / number2

else: 
    answer = "Invalid Operation"

print(answer)
```
